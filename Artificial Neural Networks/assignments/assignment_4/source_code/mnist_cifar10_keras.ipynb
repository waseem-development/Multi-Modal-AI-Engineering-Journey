# MNIST and CIFAR-10 with TensorFlow Keras

## 1. MNIST (full dataset)

### Load and prepare the data

```python
from tensorflow import keras

(trainX, trainY), (testX, testY) = keras.datasets.mnist.load_data()

# flatten each image into one vector and scale pixels to 0-1
trainX = trainX.reshape(60000, 784) / 255.0
testX = testX.reshape(10000, 784) / 255.0

print("Training data:", trainX.shape)
print("Testing data:", testX.shape)
```

### Step 1: Define the model

```python
model = keras.Sequential([
    keras.layers.Input(shape=(784,)),
    keras.layers.Dense(256, activation="relu"),
    keras.layers.Dense(128, activation="relu"),
    keras.layers.Dense(10, activation="softmax"),
])

model.summary()
```

### Step 2: Compile the model

```python
model.compile(
    optimizer=keras.optimizers.SGD(learning_rate=0.01, momentum=0.9),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)
```

### Step 3: Fit (train) the model

```python
history = model.fit(
    trainX, trainY,
    validation_data=(testX, testY),
    epochs=10,
    batch_size=128,
)
```

### Step 4: Evaluate the model

```python
loss, accuracy = model.evaluate(testX, testY)
print(f"MNIST test loss = {loss:.4f}")
print(f"MNIST test accuracy = {accuracy * 100:.2f}%")
```

### Step 5: Predict

```python
probabilities = model.predict(testX[:5])

print("Predicted classes:", probabilities.argmax(axis=1))
print("Actual classes:   ", testY[:5].flatten())
```

## 2. CIFAR-10

### Load and prepare the data

```python
from tensorflow import keras

(trainX, trainY), (testX, testY) = keras.datasets.cifar10.load_data()

# flatten each image into one vector and scale pixels to 0-1
trainX = trainX.reshape(50000, 3072) / 255.0
testX = testX.reshape(10000, 3072) / 255.0

print("Training data:", trainX.shape)
print("Testing data:", testX.shape)
```

### Step 1: Define the model

```python
model = keras.Sequential([
    keras.layers.Input(shape=(3072,)),
    keras.layers.Dense(512, activation="relu"),
    keras.layers.Dense(256, activation="relu"),
    keras.layers.Dense(10, activation="softmax"),
])

model.summary()
```

### Step 2: Compile the model

```python
model.compile(
    optimizer=keras.optimizers.SGD(learning_rate=0.01, momentum=0.9),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)
```

### Step 3: Fit (train) the model

```python
history = model.fit(
    trainX, trainY,
    validation_data=(testX, testY),
    epochs=20,
    batch_size=128,
)
```

### Step 4: Evaluate the model

```python
loss, accuracy = model.evaluate(testX, testY)
print(f"CIFAR-10 test loss = {loss:.4f}")
print(f"CIFAR-10 test accuracy = {accuracy * 100:.2f}%")
```

### Step 5: Predict

```python
probabilities = model.predict(testX[:5])

print("Predicted classes:", probabilities.argmax(axis=1))
print("Actual classes:   ", testY[:5].flatten())
```
