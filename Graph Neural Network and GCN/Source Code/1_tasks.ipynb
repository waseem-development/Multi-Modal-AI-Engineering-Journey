```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.Graph()

G.add_nodes_from(["Qasim Jungli", "Hashim Baboon", "Ashir Racoon", "Faris Fox", "Hareem"])

G.add_edges_from([
    ("Qasim Jungli", "Hashim Baboon"), 
    ("Hashim Baboon", "Ashir Racoon"), 
    ("Ashir Racoon", "Faris Fox"), 
    ("Faris Fox", "Hareem"), 
    ("Hareem", "Qasim Jungli")
])

nx.draw(G, with_labels=True)
plt.show()
```

```python
G.add_nodes_from(["Waseem", "Naruto"])
G.add_edges_from([["Waseem", "Naruto"], ["Waseem", 
"Ashir Racoon"], ["Naruto", "Faris Fox"]])
nx.draw(G, with_labels=True)
```

```python
G.remove_node("Hashim Baboon")
nx.draw(G, with_labels=True)
```

```python
from networkx.readwrite import json_graph as jsg

json_data = jsg.node_link_data(G)

json_data
```

```python
json_to_graph = jsg.node_link_graph(json_data)
nx.draw(json_to_graph, with_labels=True)
plt.title("Graph G")
plt.show()
```

### *Directional Graph*

```python
di_graph = nx.DiGraph()

di_graph.add_nodes_from(["A", "B", "C", "D"])

di_graph.add_edges_from([
    ("A", "B", {"weight": 1}),
    ("B", "C", {"weight": 10}),
    ("C", "D", {"weight": 100}),
    ("D", "A", {"weight": 1000})
])

nx.draw(di_graph, 
        node_color="#A3FF00",
        with_labels=True,
        node_size=1500,
        arrowsize=20
        )
plt.title("Directional Graph")
plt.show()
```

```python
print(di_graph.is_directed())  # True
print(di_graph.number_of_nodes())
print(di_graph.number_of_edges())
print(di_graph.edges())
print(di_graph.graph)

di_graph.graph["name"] = "My Directed Graph"
di_graph.graph["description"] = "A simple directed graph"
di_graph.graph["created_by"] = "Paxto"


nx.draw(di_graph, 
        node_color="#A3FF00",
        with_labels=True,
        node_size=1500,
        arrowsize=20
        )

plt.title("My Directed Graph")
plt.show()
```

```python
print(f"nx.is_weighted(di_graph): {nx.is_weighted(di_graph)}")
print(f"nx.is_directed(di_graph): {nx.is_directed(di_graph)}")
```
