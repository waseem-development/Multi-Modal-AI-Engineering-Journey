```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

# Create an undirected graph
G = nx.Graph()

# Add edges
G.add_edges_from([
    ("A", "B"),
    ("A", "C"),
    ("B", "D"),
    ("B", "E"),
    ("C", "D"),
    ("D", "E"),
    ("E", "F")
])

# Draw the graph
nx.draw(
    G,
    with_labels=True,
    node_color="skyblue",
    node_size=1000,
    font_size=14,
    font_weight="bold"
)

plt.show()
```

```python
A = nx.adjacency_matrix(G)
print(A.todense())
A_num = A.toarray()
print(A_num)
```

```python
degrees = A_num.sum(axis=1)
print(degrees)
```

```python
for node in G.nodes():
    print(f"Negighbors of node {node} are {list(G.neighbors(node))}")
print(f"Node ordering for graph G is: {list(G.nodes())}")
```

```python
print(np.array_equal(A_num, A_num.T))
```

### `Graph Density`
Graph density tells us how many edges actually exist compared with how many edges could possibly exist.

$$
D = \frac{2E}{N(N-1)}
$$


```python
nx.density(G)
```

Above code output means that nearly half of all possible connections exist in the graph G.

```python
from networkx.classes.function import degree

degree(G, nbunch=None, weight=None)

# nbunch: Which node or nodes you want degrees for

# nbunch=None means calculate the degree of every node.
```

```python
print(dict(nx.degree(G))) # same as above
```

```python
print(dict(nx.degree(G, ['A', 'B', 'C'])))
```

```python
G.add_edge("A", "B", weight=5)
G.add_edge("A", "C", weight=10)
# Normal degree = Number of connections.
print(G.degree("A"))
print(G.degree("A", weight="weight"))
# Weighted degree = Sum of edge weights.
```

```python
from networkx.classes.function import degree_histogram

degree_histogram(G)
```

```python
from networkx.classes.function import neighbors

print(list(neighbors(G, "A")))
```

```python
from networkx.classes.function import all_neighbors

list(all_neighbors(G, "A"))
```

- `neighbors()`	Outgoing neighbors in a directed graph\\
- `all_neighbors()`	Both incoming and outgoing neighbors
